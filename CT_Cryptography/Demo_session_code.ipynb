```python
# Code 1: Find the Order of an Element

def find_order(g, N):
    value = 1
    for k in range(1, N + 1):
        value = (value * g) % N
        if value == 1:
            return k

print("Order of 2 mod 5:", find_order(2, 5))
print("Order of 3 mod 5:", find_order(3, 5))
```

```python
# Code 2: Find Primitive Roots

import math

def primitive_roots(N):
    roots = []

    for g in range(1, N):
        if math.gcd(g, N) == 1:
            if find_order(g, N) == sum(
                1 for i in range(1, N) if math.gcd(i, N) == 1
            ):
                roots.append(g)

    return roots

print("Primitive roots modulo 5:", primitive_roots(5))
```

```python
# Code 3: Diffie-Hellman Key Exchange

N = 11
g = 7

a = 2
b = 4

A = pow(g, a, N)
B = pow(g, b, N)

print("Alice's public value:", A)
print("Bob's public value:", B)

k1 = pow(A, b, N)
k2 = pow(B, a, N)

print("Alice's shared key:", k1)
print("Bob's shared key:", k2)

if k1 == k2:
    print("The shared synthesized key is", k1)
else:
    print("Check if the generator is valid")
```

```python
# Code 4: Check Prime Numbers

import math

def CheckIfPrime(n):
    if n < 2:
        return False

    for i in range(2, int(math.sqrt(n)) + 1):
        if n % i == 0:
            return False

    return True

p = 5
q = 11

print("Is p prime?", CheckIfPrime(p))
print("Is q prime?", CheckIfPrime(q))
```

```python
# Code 5: Calculate N and Euler Totient Function

p = 5
q = 11

N = p * q
phiN = (p - 1) * (q - 1)

print("N =", N)
print("Euler Totient Function =", phiN)
```

```python
# Code 6: Extended Euclidean Algorithm

def extended_gcd(a, b):
    if b == 0:
        return a, 1, 0

    g, x1, y1 = extended_gcd(b, a % b)

    x = y1
    y = x1 - (a // b) * y1

    return g, x, y

g, x, y = extended_gcd(7, 40)

print("GCD =", g)
print("x =", x)
print("y =", y)
```

```python
# Code 8: RSA Encryption

def Encrypt(e, N, message):
    return pow(message, e, N)

M = 2

ciphertext = Encrypt(e, N, M)

print("Plaintext:", M)
print("Ciphertext:", ciphertext)
```

```python
# Code 9: RSA Decryption

def Decrypt(d, N, ciphertext):
    return pow(ciphertext, d, N)

decryptedtext = Decrypt(d, N, ciphertext)

print("Decrypted text:", decryptedtext)
```

```python
# Code 10: Check Relatively Prime Numbers

import math

N = 35
a = 8

print("GCD of a and N:", math.gcd(a, N))

if math.gcd(a, N) == 1:
    print("a and N are relatively prime")
else:
    print("Choose another value of a")
```

```python
# Code 11: Find the Period

def find_period(a, N):
    value = 1

    for r in range(1, N + 1):
        value = (value * a) % N

        if value == 1:
            return r

    return None

r = find_period(a, N)

print("Period =", r)
```

```python
# Code 12: Calculate the Factors

if r is not None and r % 2 == 0:
    x = pow(a, r // 2, N)

    print("a^(r/2) mod N =", x)

    if x != 1 and x != N - 1:
        factor1 = math.gcd(x - 1, N)
        factor2 = math.gcd(x + 1, N)

        print("First factor:", factor1)
        print("Second factor:", factor2)
    else:
        print("Choose another value of a")
else:
    print("Period is odd or not found. Choose another a.")
```

```python
# Code 13: Verify the Factors

print("Original number:", N)
print("Product of factors:", factor1 * factor2)

if factor1 * factor2 == N:
    print("Factorization successful!")
else:
    print("Factorization unsuccessful!")
```
