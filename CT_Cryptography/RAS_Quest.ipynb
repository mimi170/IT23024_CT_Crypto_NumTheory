```python
import sys
import threading
sys.setrecursionlimit(10**7)
threading.stack_size(2**27)

def ConvertToInt(message_str):
    res = 0
    for i in range(len(message_str)):
        res = res * 256 + ord(message_str[i])
    return res

def ConvertToStr(n):
    res = ""
    while n > 0:
        res += chr(n % 256)
        n //= 256
    return res[::-1]

# Question 1: PowMod
def PowMod(a, n, mod):
    if n == 0:
        return 1 % mod
    elif n == 1:
        return a % mod
    else:
        b = PowMod(a, n // 2, mod)
        b = b * b % mod
        if n % 2 == 0:
            return b
        else:
            return b * a % mod

# Question 2: Extended Euclid
def ExtendedEuclid(a, b):
    if b == 0:
        return (1, 0)
    (x, y) = ExtendedEuclid(b, a % b)
    k = a // b
    return (y, x - k * y)

# Question 3: InvertModulo
def InvertModulo(a, n):
    (b, x) = ExtendedEuclid(a, n)
    if b < 0:
        b = (b % n + n) % n
    return b

# Question 4: RSA Decryption
def Decrypt(ciphertext, p, q, exponent):
    modulo = p * q
    phi = (p - 1) * (q - 1)
    d = InvertModulo(exponent, phi)
    m = PowMod(ciphertext, d, modulo)
    return ConvertToStr(m)

p = 779849711281
q = 748173698927
e = 1018651
ciphertext = 148784435264686331994392

decrypt_first_puzzle = Decrypt(ciphertext, p, q, e)
print(decrypt_first_puzzle)
```

```python
# Question 1: RSA Encryption
def Encrypt(message, modulo, exponent):
    a = ConvertToInt(message)
    return PowMod(a, exponent, modulo)

# Question 3: DecipherSimple
def DecipherSimple(ciphertext, modulo, exponent, potential_messages):
    for message in potential_messages:
        encrypted = Encrypt(message, modulo, exponent)

        if encrypted == ciphertext:
            return message

    return "don't know"

ciphertext = 336184023047118677086739
modulo = 1110014195838866450995043
exponent = 767549

potential_messages = [
    "http://goo.gl/",
    "http://bit.ly/",
    "http://tinyurl.com/",
    "http://t.co/",
    "http://ow.ly/",
    "https://is.gd/",
    "https://buff.ly/",
    "http://adf.ly/",
    "http://bit.do/"
]

decrypt_second_puzzle = DecipherSimple(
    ciphertext, modulo, exponent, potential_messages
)

print(decrypt_second_puzzle)
```

```python
# Question 4: DecipherSmallPrime

def DecipherSmallPrime(ciphertext, modulo, exponent):
    for p in range(2, 1000000):
        if modulo % p == 0:
            q = modulo // p
            return Decrypt(ciphertext, p, q, exponent)

    return "don't know"

ciphertext = 0
modulo = 0
exponent = 0

decrypt_third_puzzle = DecipherSmallPrime(ciphertext, modulo, exponent)
print(decrypt_third_puzzle)
```

```python
# Question 5: IntSqrt

def IntSqrt(n):
    low = 1
    high = n
    iterations = 0

    while low < high and iterations < 5000:
        iterations += 1
        mid = (low + high + 1) // 2

        if mid * mid <= n:
            low = mid
        else:
            high = mid - 1

    return low

# Question 5: DecipherSmallDiff

def DecipherSmallDiff(ciphertext, modulo, exponent):
    sqrtn = IntSqrt(modulo)

    x = sqrtn
    if x * x < modulo:
        x += 1

    while True:
        y_squared = x * x - modulo
        y = IntSqrt(y_squared)

        if y * y == y_squared:
            p = x - y
            q = x + y

            if p * q == modulo:
                return Decrypt(ciphertext, p, q, exponent)

        x += 1

ciphertext = 0
modulo = 0
exponent = 0

decrypt_fourth_puzzle = DecipherSmallDiff(ciphertext, modulo, exponent)
print(decrypt_fourth_puzzle)
```

```python
# Question 6: GCD

def GCD(a, b):
    if b == 0:
        return a
    return GCD(b, a % b)

# Question 6: DecipherCommonDivisor

def DecipherCommonDivisor(first_ciphertext, first_modulo, first_exponent,
                          second_ciphertext, second_modulo, second_exponent):

    common_prime = GCD(first_modulo, second_modulo)

    if common_prime == 1:
        return ("don't know", "don't know")

    q1 = first_modulo // common_prime
    q2 = second_modulo // common_prime

    first_message = Decrypt(
        first_ciphertext, common_prime, q1, first_exponent
    )

    second_message = Decrypt(
        second_ciphertext, common_prime, q2, second_exponent
    )

    return (first_message, second_message)



first_ciphertext = 0
first_modulo = 0
first_exponent = 0

second_ciphertext = 0
second_modulo = 0
second_exponent = 0

decrypt_sixth_puzzle = DecipherCommonDivisor(
    first_ciphertext, first_modulo, first_exponent,
    second_ciphertext, second_modulo, second_exponent
)

print(decrypt_sixth_puzzle)
```

```python
# Question 7: ChineseRemainderTheorem

def ChineseRemainderTheorem(n1, r1, n2, r2):
    inverse = InvertModulo(n1, n2)
    k = ((r2 - r1) * inverse) % n2
    return r1 + n1 * k


# Question 7: DecipherHastad

def DecipherHastad(first_ciphertext, first_modulo,
                   second_ciphertext, second_modulo):

    m_squared = ChineseRemainderTheorem(
        first_modulo, first_ciphertext,
        second_modulo, second_ciphertext
    )

    m = IntSqrt(m_squared)

    if m * m != m_squared:
        return "don't know"

    return ConvertToStr(m)


# দ্বিতীয় Secret Document থেকে মান বসাবে
first_ciphertext = 0
first_modulo = 0

second_ciphertext = 0
second_modulo = 0

decrypt_seventh_puzzle = DecipherHastad(
    first_ciphertext, first_modulo,
    second_ciphertext, second_modulo
)

print(decrypt_seventh_puzzle)
```
