```python
def solve_problem_1(max_n=40):
    print("--- Problem 1: Repunits and Primes ---")
    primes = [2, 3, 5, 7, 11, 13, 17, 19]
    for p in primes:
        ap_set = []
        for n in range(1, max_n + 1):
            # a(n) তৈরি করা (যেমন: n=3 হলে 111)
            val = int('1' * n)
            if val % p == 0:
                ap_set.append(n)
        print(f"Prime p = {p}: A_p = {ap_set}")

solve_problem_1()
```

```python
def is_prime(num):
    if num < 2: return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0: return False
    return True

def solve_problem_2(limit=10):
    print("\n--- Problem 2: Even Perfect Numbers ---")
    print("Finding even perfect numbers using formula 2^(n-1) * (2^n - 1):")

    count = 0
    n = 1
    while count < limit:
        mersenne = (2**n) - 1
        if is_prime(n) and is_prime(mersenne):
            perfect_num = (2**(n - 1)) * mersenne
            print(f"n = {n} -> Perfect Number: {perfect_num}")
            count += 1
        n += 1

solve_problem_2()
```

```python
def is_prime(num):
    if num < 2: return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0: return False
    return True

def solve_problem_3(base=2, max_pseudoprimes=5):
    print(f"\n--- Problem 3: Fermat Pseudoprimes for base a = {base} ---")
    found = 0
    n = 3
    while found < max_pseudoprimes:

        if not is_prime(n):

            if pow(base, n - 1, n) == 1:
                print(f"Found Fermat Pseudoprime: {n} (since {base}^{n-1} ≡ 1 mod {n})")
                found += 1
        n += 2

solve_problem_3(base=2)
```
