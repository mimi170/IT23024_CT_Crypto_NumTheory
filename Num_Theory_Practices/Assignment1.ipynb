```python
# Question 1: Extended Euclid's Algorithm

def extended_gcd(a, b):
    assert a >= b and b >= 0 and a + b > 0

    d, x, y = 0, 0, 0

    if b == 0:
        d = a
        x = 1
        y = 0
    else:
        d, x1, y1 = extended_gcd(b, a % b)
        x = y1
        y = x1 - (a // b) * y1

    assert a % d == 0 and b % d == 0
    assert d == a * x + b * y

    return (d, x, y)


print(extended_gcd(8, 4))
print(extended_gcd(27, 15))
```

```python
# Question 2: Least Common Multiple

def lcm(a, b):
    d, x, y = extended_gcd(max(a, b), min(a, b))
    return (a * b) // d


assert lcm(9, 3) == 9
assert lcm(17, 37) == 629

print(lcm(9, 3))
print(lcm(17, 37))
```

```python
# Question 3: Diophantine Equation

def diophantine(a, b, c):
    gcd, x1, y1 = extended_gcd(max(abs(a), abs(b)), min(abs(a), abs(b)))

    if abs(a) >= abs(b):
        if a < 0:
            x1 = -x1
        if b < 0:
            y1 = -y1
        x, y = x1, y1
    else:
        if a < 0:
            y1 = -y1
        if b < 0:
            x1 = -x1
        x, y = y1, x1

    x = x * (c // gcd)
    y = y * (c // gcd)

    assert c % gcd == 0
    assert a * x + b * y == c

    return (x, y)


print(diophantine(34, 51, 170))
print(diophantine(91, 117, 65))
```

```python
# Question 4: Modular Inverse

def modInverse(a, n):
    d, x, y = extended_gcd(max(a, n), min(a, n))

    if a > n:
        return x % n
    else:
        return y % n


def divide(b, a, n):
    assert n > 1 and a > 0

    aInv = modInverse(a, n)

    assert (a * aInv) % n == 1

    return (aInv * b) % n


print(divide(9, 5, 3))
print(divide(223, 67, 54))
```

```python
# Question 5: Fast Modular Exponentiation

def FastModularExponentiation(b, e, m):

    result = 1
    b = b % m

    while e > 0:
        if e % 2 == 1:
            result = (result * b) % m

        b = (b * b) % m
        e = e // 2

    return result


print(FastModularExponentiation(4, 5, 3))
```

```python
# Question 6: Chinese Remainder Theorem

def ChineseRemainderTheorem(n1, r1, n2, r2):

    (d, x, y) = extended_gcd(n1, n2)

    assert d == 1

    result = r1 * n2 * y + r2 * n1 * x

    return result % (n1 * n2)


print(ChineseRemainderTheorem(13, 7, 6, 5))  # 59
print(ChineseRemainderTheorem(28, 20, 19, 3))  # 440
```
